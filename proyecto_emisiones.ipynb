```python
# ==========================================
# PROYECTO FINAL: Analisis de Emisiones de CO2
# Carga y Limpieza de Datos
# ==========================================

import numpy as np
import pandas as pd

# 1. Carga del dataset oficial de Our World in Data
url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
print("Descargando dataset desde Our World in Data...")
df = pd.read_csv(url)

print(f"Dimensiones originales del DataFrame: {df.shape}")

# 2. Seleccion de variables clave para el analisis
columnas_interes = ['country', 'year', 'co2', 'co2_per_capita', 'population']
df_filtrado = df[columnas_interes].copy()

# 3. Filtrar para un enfoque regional o pais especifico (Ejemplo: Colombia)
paises_objetivo = ['Colombia', 'World']
df_limpio = df_filtrado[df_filtrado['country'].isin(paises_objetivo)].copy()

# Filtrar un rango de anos consistente (desde 1950 en adelante) y eliminar nulos en 'co2'
df_limpio = df_limpio[(df_limpio['year'] >= 1950) & (df_limpio['co2'].notna())]

# 4. Verificacion de limpieza
print("\n--- Informacion del Dataset Limpio ---")
print(df_limpio.info())

print("\n--- Primeros registros ---")
display(df_limpio.head())

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Evolución de emisiones totales de CO2 (1950 - Presente)
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=df_limpio,
    x="year",
    y="co2",
    hue="country",
    marker="o",
    linewidth=2.2,
)
plt.title(
    "Evolución de Emisiones Totales de CO₂ (1950 - Presente): Colombia vs Mundo",
    fontsize=13,
    weight="bold",
)
plt.xlabel("Año", fontsize=11)
plt.ylabel("Emisiones de CO₂ (Millones de toneladas)", fontsize=11)
plt.legend(title="Entidad")
plt.tight_layout()
plt.show()

# 2. Evolución de emisiones de CO2 per cápita (1950 - Presente)
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=df_limpio,
    x="year",
    y="co2_per_capita",
    hue="country",
    marker="s",
    linewidth=2.2,
)
plt.title(
    "Emisiones de CO₂ Per Cápita (1950 - Presente): Colombia vs Mundo",
    fontsize=13,
    weight="bold",
)
plt.xlabel("Año", fontsize=11)
plt.ylabel("CO₂ Per Cápita (Toneladas por persona)", fontsize=11)
plt.legend(title="Entidad")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

sns.set_theme(style="whitegrid")

# Filtramos los datos de Colombia desde 1950 en adelante
df_col = df_limpio[df_limpio["country"] == "Colombia"].dropna(subset=["year", "co2"]).copy()

x = df_col["year"].to_numpy(dtype=float)
y = df_col["co2"].to_numpy(dtype=float)

# Ajuste de regresión lineal: y = pendiente * x + intercepto
pendiente, intercepto = np.polyfit(x, y, deg=1)
y_pred = pendiente * x + intercepto

# Métricas del modelo: R^2 y RMSE
ss_res = np.sum((y - y_pred) ** 2)
ss_tot = np.sum((y - np.mean(y)) ** 2)
r2 = 1 - (ss_res / ss_tot)
rmse = np.sqrt(np.mean((y - y_pred) ** 2))

print(f"Pendiente (cambio anual promedio): {pendiente:.4f} Mt CO2/año")
print(f"Intercepto: {intercepto:.4f}")
print(f"Coeficiente de determinación (R²): {r2:.4f}")
print(f"Error cuadrático medio (RMSE): {rmse:.4f} Mt CO2")

# Gráfica de dispersión + línea de tendencia lineal
plt.figure(figsize=(11, 5.5))
sns.scatterplot(x=x, y=y, label="Datos observados (Colombia)", alpha=0.75)
plt.plot(x, y_pred, color="crimson", linewidth=2.2, label=f"Regresión lineal (R² = {r2:.3f})")
plt.title("Modelo de Regresión Lineal — Emisiones de CO₂ (Colombia)", fontsize=13, weight="bold")
plt.xlabel("Año", fontsize=11)
plt.ylabel("Emisiones de CO₂ (Millones de toneladas)", fontsize=11)
plt.legend()
plt.tight_layout()
plt.show()
```

## Conclusiones del Proyecto 

- **Tendencia en Colombia:** El modelo de regresión lineal muestra una pendiente de **1.2503 Mt CO₂/año**, evidenciando un crecimiento promedio sostenido de las emisiones de CO₂ durante el periodo analizado.
- **Bondad de ajuste:** Se obtuvo un coeficiente de determinación **$R^2$ de 0.9626** y un **RMSE de 5.4047 Mt CO₂**, lo que indica un buen ajuste general de la tendencia lineal, aunque con variaciones puntuales entre periodos que una recta simple no captura por sí sola.
- **Cierre del análisis:** Combinar la limpieza de datos, la visualización comparativa y la regresión lineal permite resumir la evolución histórica de las emisiones y deja una base clara para profundizar en otros países, variables o modelos.
