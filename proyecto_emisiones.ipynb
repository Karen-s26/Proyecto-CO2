```python
# ==========================================
# PROYECTO FINAL: Analisis de Emisiones de CO2
# Carga y Limpieza de Datos
# ==========================================

import numpy as np
import pandas as pd

# 1. Carga del dataset oficial de Our World in Data
url = "https://raw.githubusercontent.com/owid/co2-data/master/owid-co2-data.csv"
print("Descargando dataset desde Our World in Data...")
df = pd.read_csv(url)

print(f"Dimensiones originales del DataFrame: {df.shape}")

# 2. Seleccion de variables clave para el analisis
columnas_interes = ['country', 'year', 'co2', 'co2_per_capita', 'population']
df_filtrado = df[columnas_interes].copy()

# 3. Filtrar para un enfoque regional o pais especifico (Ejemplo: Colombia)
paises_objetivo = ['Colombia', 'World']
df_limpio = df_filtrado[df_filtrado['country'].isin(paises_objetivo)].copy()

# Filtrar un rango de anos consistente (desde 1950 en adelante) y eliminar nulos en 'co2'
df_limpio = df_limpio[(df_limpio['year'] >= 1950) & (df_limpio['co2'].notna())]

# 4. Verificacion de limpieza
print("\n--- Informacion del Dataset Limpio ---")
print(df_limpio.info())

print("\n--- Primeros registros ---")
display(df_limpio.head())

```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

# 1. Evolución de emisiones totales de CO2 (1950 - Presente)
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=df_limpio,
    x="year",
    y="co2",
    hue="country",
    marker="o",
    linewidth=2.2,
)
plt.title(
    "Evolución de Emisiones Totales de CO₂ (1950 - Presente): Colombia vs Mundo",
    fontsize=13,
    weight="bold",
)
plt.xlabel("Año", fontsize=11)
plt.ylabel("Emisiones de CO₂ (Millones de toneladas)", fontsize=11)
plt.legend(title="Entidad")
plt.tight_layout()
plt.show()

# 2. Evolución de emisiones de CO2 per cápita (1950 - Presente)
plt.figure(figsize=(12, 6))
sns.lineplot(
    data=df_limpio,
    x="year",
    y="co2_per_capita",
    hue="country",
    marker="s",
    linewidth=2.2,
)
plt.title(
    "Emisiones de CO₂ Per Cápita (1950 - Presente): Colombia vs Mundo",
    fontsize=13,
    weight="bold",
)
plt.xlabel("Año", fontsize=11)
plt.ylabel("CO₂ Per Cápita (Toneladas por persona)", fontsize=11)
plt.legend(title="Entidad")
plt.tight_layout()
plt.show()
```
